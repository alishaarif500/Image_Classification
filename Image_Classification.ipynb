# Project Plan & Architectural Overview: Dogs vs. Cats Image Classification

## 1. Introduction and Dataset Justification
For this research project, we select the academic benchmark **Kaggle Dogs vs. Cats** dataset. This dataset is widely recognized in the computer vision community and provides an optimal testbed for evaluating custom Convolutional Neural Network (CNN) architectures and Transfer Learning paradigms.

### Dataset Specifications:
- **Structure**: The dataset comprises two main categories: `Cat` and `Dog`.
- **Size**: Contains 25,000 labeled training images of dogs and cats (12,500 of each class).
- **Format**: Standard JPEG images of varying resolutions, aspect ratios, and orientations, requiring a robust preprocessing pipeline.
- **Licensing & Accessibility**: Publicly accessible via Kaggle for educational and academic research purposes.
- **Academic Value**: Because of its realistic variations in background, lighting, and animal poses, it serves as a highly effective benchmark to study model generalization and prevent overfitting.

---

## 2. Research Objectives
Our research aims to analyze the efficiency-performance trade-offs in modern deep learning models for binary image classification. Specifically:
- **Primary Objective**: Design, train, and evaluate a baseline custom Convolutional Neural Network (CNN) and compare it against a state-of-the-art Transfer Learning model (e.g., ResNet50 or MobileNetV2).
- **Target Metrics**:
  - **Accuracy**: Achieve $\ge 90\%$ validation accuracy using Transfer Learning.
  - **Loss**: Minimize Cross-Entropy Loss on the validation set.
  - **Robustness**: Evaluate Precision, Recall, F1-Score, and Area Under the ROC Curve (ROC-AUC) to ensure balanced class performance.

---

## 3. Deep Learning System Architecture
To ensure modularity and scalability, the system is designed in four major pipeline stages:

```
+-----------------------+     +--------------------------+     +------------------------+     +------------------------+
| 1. Data Preprocessing | --> | 2. Model Architecture     | --> | 3. Training Strategy   | --> | 4. Evaluation Methods  |
| - Resize to 224x224   |     | - Option A: Custom CNN   |     | - Optimizers: Adam/SGD |     | - Accuracy & F1-Score  |
| - Rescale [0, 1]      |     | - Option B: Transfer     |     | - Loss: Cross-Entropy  |     | - Confusion Matrix     |
| - Data Augmentation  |     |   Learning (ResNet50)    |     | - Callbacks: Early     |     | - ROC-AUC Curves       |
|   (Flip, Rotation)    |     |                          |     |   Stopping, ReduceLR   |     |                        |
+-----------------------+     +--------------------------+     +------------------------+     +------------------------+
```

### 3.1 Data Preprocessing & Augmentation
- **Resizing**: Standardize all input images to $224 \times 224 \times 3$ pixels to match the input requirements of pretrained networks.
- **Normalization**: Rescale pixel intensity values from $[0, 255]$ to $[0, 1]$ or normalize using ImageNet mean and standard deviation.
- **Data Augmentation**: Apply real-time transformations to the training set (random horizontal flips, rotations up to $15^\circ$, zooming, and shearing) to improve generalization.

### 3.2 Model Selection
1. **Baseline Custom CNN**: A lightweight architecture featuring sequential `Conv2D`, `BatchNormalization`, `MaxPooling2D`, and `Dropout` layers, culminating in a `Dense` output layer with Sigmoid activation.
2. **Transfer Learning Model**: Leveraging a pre-trained **ResNet50** or **VGG16** backbone trained on ImageNet. We will freeze the initial feature extraction blocks and attach a custom dense classifier head for binary categorization.

### 3.3 Training Strategy
- **Loss Function**: Binary Cross-Entropy Loss.
- **Optimizers**: Adam (with an initial learning rate of $1e-4$) or SGD with Momentum.
- **Callbacks**:
  - `EarlyStopping` to monitor validation loss and halt training if overfitting occurs.
  - `ReduceLROnPlateau` to dynamically reduce the learning rate when validation loss plateaus.

### 3.4 Evaluation Methods
- Multi-dimensional evaluation using a distinct validation and test set.
- Detailed performance reports including **Confusion Matrices**, **Classification Reports (Precision, Recall, F1-Score)**, and **ROC-AUC curves** to visually inspect binary decision thresholds.

# Project Plan & Architecture: Dogs vs. Cats Image Classification

## 1. Dataset Selection & Justification
We select the academic standard **Kaggle Dogs vs. Cats** dataset. This dataset consists of thousands of labeled images of dogs and cats.

### Academic Justification:
* **Real-World Variability:** Images exhibit diverse backgrounds, lighting conditions, angles, scales, and biological variations of breeds.
* **Binary Classification Benchmark:** Serves as a fundamental computer vision benchmark to compare custom CNN architectures against established pre-trained models.
* **Generalization Challenge:** Successfully separating these two similar quadruped classes requires extracting both coarse structural features and fine grain textures.

---

## 2. Research Objectives & Target Performance Metrics
Our goals are to design, implement, and compare two distinct deep learning models:
1. **A Custom CNN Architecture:** Designed from scratch to understand baseline feature representation learning.
2. **A Transfer Learning Model (e.g., MobileNetV2 or ResNet50):** Leveraging features pre-trained on ImageNet.

### Target Metrics:
* **Validation Accuracy:** Target $\ge 80\%$ for Custom CNN, and $\ge 95\%$ for Transfer Learning.
* **Loss:** Low Cross-Entropy Loss on validation split with minimal overfitting.
* **F1-Score / Precision / Recall:** Comprehensive evaluation to confirm balanced class predictions across both categories.

---

## 3. Deep Learning Pipeline Architecture
The system is organized into five modular stages:

```
[ Raw Data ]
     │
     ▼
[ Preprocessing & Augmentation ] (Resize to 224x224, Normalization, Random Rotation, Flips)
     │
     ├─────────────────────────────────────────┐
     ▼                                         ▼
[ Custom CNN Pipeline ]             [ Transfer Learning Pipeline ]
  - Conv2D + BatchNorm + ReLU         - Pre-trained Base (Frozen backbone)
  - MaxPool2D + Dropout               - Custom Dense / Classification Head
     │                                         │
     └────────────────────┬────────────────────┘
                          ▼
                [ Training Strategy ] (Adam Optimizer, Early Stopping, ReduceLROnPlateau)
                          │
                          ▼
                [ Evaluation & Diagnostics ] (Confusion Matrix, ROC Curve, Epoch Curves)
```

### Pipeline Modules:
1. **Preprocessing & Augmentation:** Input normalization, resizing to $224 \times 224$, and on-the-fly augmentation (rotation, zooming, horizontal flips) to prevent overfitting.
2. **Custom CNN:** Deep convolutional layers with Batch Normalization, Dropout for regularization, and MaxPooling.
3. **Transfer Learning:** Using a robust feature extractor (e.g., MobileNetV2) coupled with a tailored classification head.
4. **Training Strategy:** Optimized using Adam/SGD optimizer, binary cross-entropy loss, dynamic learning rate schedulers (`ReduceLROnPlateau`), and `EarlyStopping` callbacks.
5. **Evaluation & Diagnostics:** Extensive diagnostic plots containing loss/accuracy curves, confusion matrices, and ROC curves.
