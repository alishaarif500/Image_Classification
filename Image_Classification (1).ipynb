## 3. Deep Learning System Architecture
To ensure modularity and scalability, the system is designed in four major pipeline stages:

```
+-----------------------+     +--------------------------+     +------------------------+     +------------------------+
| 1. Data Preprocessing | --> | 2. Model Architecture     | --> | 3. Training Strategy   | --> | 4. Evaluation Methods  |
```

### 3.1 Data Preprocessing & Augmentation

- **Normalization**: Rescale pixel intensity values from $[0, 255]$ to $[0, 1]$ or normalize using ImageNet mean and standard deviation.
- **Data Augmentation**: Apply real-time transformations to the training set (random horizontal flips, rotations up to $15^\circ$, zooming, and shearing) to improve generalization.

### 3.3 Training Strategy

- **Optimizers**: Adam (with an initial learning rate of $1e-4$) or SGD with Momentum.
- **Callbacks**:
  - `EarlyStopping` to monitor validation loss and halt training if overfitting occurs.

### 3.4 Evaluation Methods
- Multi-dimensional evaluation using a distinct validation and test set.

### Dataset Filtering & Splitting
Now, we split the raw images into clean train, validation, and test sets. We will filter out corrupted images (e.g., zero-byte files) to ensure stable model training.

````python
import os
import shutil
import random

def clean_and_split_data(src_dir, train_dir, val_dir, test_dir, split_ratio=(0.8, 0.1, 0.1)):
    all_files = []
    for root, _, files in os.walk(src_dir):
        for file in files:
            if file.lower().endswith(('.jpg', '.jpeg')):
                path = os.path.join(root, file)
                # Filter out empty or corrupted images
                if os.path.getsize(path) > 1024:
                    all_files.append(path)

    # Group by class
    cats = [f for f in all_files if 'cat' in os.path.basename(f).lower() or 'cat' in f.lower()]
    dogs = [f for f in all_files if 'dog' in os.path.basename(f).lower() or 'dog' in f.lower()]

    random.seed(42)
    random.shuffle(cats)
    random.shuffle(dogs)

    # Limit to 1000 images per class for execution speed in this environment
    cats = cats[:1000]
    dogs = dogs[:1000]

    train_idx = int(len(cats) * split_ratio[0])
    val_idx = train_idx + int(len(cats) * split_ratio[1])

    splits = {
        'train/cats': cats[:train_idx],
        'train/dogs': dogs[:train_idx],
        'val/cats': cats[train_idx:val_idx],
        'val/dogs': dogs[train_idx:val_idx],
        'test/cats': cats[val_idx:],
        'test/dogs': dogs[val_idx:]
    }

    for split_name, files in splits.items():
        dest_path = f"data/processed/{split_name}"
        os.makedirs(dest_path, exist_ok=True)
        for f in files:
            shutil.copy(f, os.path.join(dest_path, os.path.basename(f)))

clean_and_split_data('data/raw/PetImages', 'data/processed/train', 'data/processed/val', 'data/processed/test')
print("Dataset split successfully!")
````

````python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Image generators with augmentation
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True
)

val_test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    'data/processed/train',
    target_size=(150, 150),
    batch_size=32,
    class_mode='binary'
)

val_generator = val_test_datagen.flow_from_directory(
    'data/processed/val',
    target_size=(150, 150),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)

test_generator = val_test_datagen.flow_from_directory(
    'data/processed/test',
    target_size=(150, 150),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)
````

````python
from tensorflow.keras import layers, models

# 1. Custom CNN Model definition
custom_model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.BatchNormalization(),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(1, activation='sigmoid')
])

custom_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train Custom CNN
print("Training Custom CNN...")
custom_history = custom_model.fit(
    train_generator,
    epochs=3,
    validation_data=val_generator
)
custom_model.save('models/custom_cnn/model.h5')
````

````python
from tensorflow.keras.applications import MobileNetV2

# 2. Transfer Learning Model using MobileNetV2
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(150, 150, 3))
base_model.trainable = False

transfer_model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')
])

transfer_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train Transfer Learning model
print("Training Transfer Learning model...")
transfer_history = transfer_model.fit(
    train_generator,
    epochs=3,
    validation_data=val_generator
)
transfer_model.save('models/transfer_learning/model.h5')
````

````python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# Evaluate and generate metrics
def generate_evaluation_artifacts(model, name):
    predictions = model.predict(test_generator)
    y_pred = (predictions > 0.5).astype(int).ravel()
    y_true = test_generator.classes

    # Confusion Matrix
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5,4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Cats', 'Dogs'], yticklabels=['Cats', 'Dogs'])
    plt.title(f'{name} Confusion Matrix')
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.savefig(f'outputs/plots/{name.lower().replace(" ", "_")}_cm.png')
    plt.close()

    # Print Report
    print(f"\n=== {name} Classification Report ===")
    print(classification_report(y_true, y_pred, target_names=['Cats', 'Dogs']))

generate_evaluation_artifacts(custom_model, "Custom CNN")
generate_evaluation_artifacts(transfer_model, "Transfer Learning")
print("Evaluation complete! All diagnostic models and plots saved successfully.")
````
